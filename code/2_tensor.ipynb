```python
d = torch.tensor([[0],
                 [1],
                 [2]])
```

```python
e = d.squeeze()
print(e)
```

```python
f = torch.tensor([[2,3,4,6],
                  [4,5,6,7],
                  [5,6,7,8],
                  [6,7,8,9]])
```

```python
# テンソル全体の平均値 (floatとして取得)
print(f.float().mean().item())

# 0次元方向（列ごと）の平均値
print(f.float().mean(0))
```

```python
a = torch.tensor([[1,2,3],
                 [4,5,6]])
b = torch.tensor([1,2,3])

print(a + b)
print(a - b)
print(a*b)
print(a/b)
print(a//b)
print(a%b)
```
