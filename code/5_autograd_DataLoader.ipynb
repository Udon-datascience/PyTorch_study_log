第４章　自動微分とDataLoader

```python
import torch
x = torch.ones(2,3,requires_grad=True)
print(x)
```

```python
y = x + 2
print(y)
print(y.grad_fn)
```

```python
z = y*3
print(z)
out = z.mean()
print(out)
```

```python
a = torch.tensor([1, 0], requires_grad=True)
b = a*2
b.backward()    #逆伝播
print(a.grad)    #aの勾配を表示
```

```python
#複雑な逆勾配の計算
def calc(a):
    b = a*2 + 1
    c = b**2
    d = c / (c + 2)
    e = d.mean()
    return e

x = [1.0, 2.0, 3.0]
x = torch.tensor(x, requires_grad=True)
y = calc(x)
y.backward()
print(x.grad)    #xの勾配
```
